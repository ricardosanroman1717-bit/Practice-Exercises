## Exercise 2: Student Grade Classifier

This exercise uses Python to assign letter grades and GPA points to student test scores.

**Grade Scale:**
- A: 90–100 (GPA 4.0)
- B: 80–89 (GPA 3.0)
- C: 70–79 (GPA 2.0)
- D: 60–69 (GPA 1.0)
- F: Below 60 (GPA 0.0)


```python

scores = [95.0, 83.5, 72.0, 61.5, 48.0, 90.0, 77.5, 55.0]

for i, score in enumerate(scores, 1):
    if score >= 90:
        grade = "A"
        gpa = 4.0
    elif score >= 80:
        grade = "B"
        gpa = 3.0
    elif score >= 70:
        grade = "C"
        gpa = 2.0
    elif score >= 60:
        grade = "D"
        gpa = 1.0
    else:
        grade = "F"
        gpa = 0.0

    print(f"#{i} | Score: {score:.2f} | Grade: {grade} | GPA: {gpa:.2f}")
```

```python

class_avg = sum(scores) / len(scores)
highest = max(scores)

print(f"Class Average: {class_avg:.2f}")
print(f"Highest Score: {highest:.2f}")
```


## Summary

The class average was **72.81**, which is in the C grade range.

The highest score was **95.00**, and two students received failing grades.



## Exercise 3: Monthly Sales Dashboard

This exercise analyzes monthly sales for five employees and calculates bonuses based on their performance.



### Data — Sales Records


```python

sales_data = [
    ("Maya", 6200.0),
    ("Leo", 3450.0),
    ("Sam", 2100.0),
    ("Priya", 5800.0),
    ("Kai", 4200.0)
]

print(f"{len(sales_data)} reps loaded.")
```


### Analysis — Performance Classification


```python

for name, sales in sales_data:
    if sales >= 5000:
        rating = "Excellent"
        bonus = sales * 0.10
    elif sales >= 3000:
        rating = "Good"
        bonus = sales * 0.05
    else:
        rating = "Needs Improvement"
        bonus = 0.0

    print(f"{name} | Sales: ${sales:.2f} | {rating} | Bonus: ${bonus:.2f}")
```


### Summary — Team Totals


```python

total = sum(sales for name, sales in sales_data)
avg = total / len(sales_data)
best = max(sales for name, sales in sales_data)

print(f"Team Total: ${total:.2f}")
print(f"Average per Rep: ${avg:.2f}")
print(f"Best Rep Sales: ${best:.2f}")
```


## Exercise 4: Temperature Data Analyser

This exercise converts Celsius temperatures to Fahrenheit and classifies different weather conditions.


```python

celsius_readings = [-5.0, 2.5, 14.0, 22.5, 30.0, 38.5]

print(f"{len(celsius_readings)} readings loaded.")
```

```python

for i, celsius in enumerate(celsius_readings, 1):
    fahrenheit = (celsius * 9/5) + 32

    if celsius < 0:
        category = "Freezing"
    elif celsius < 15:
        category = "Cold"
    elif celsius < 30:
        category = "Warm"
    else:
        category = "Hot"

    print(f"#{i} | Celsius: {celsius:.2f} | Fahrenheit: {fahrenheit:.2f} | {category}")
```

```python

avg_temp = sum(celsius_readings) / len(celsius_readings)
hottest = max(celsius_readings)
coldest = min(celsius_readings)

print(f"Average: {avg_temp:.2f}°C")
print(f"Hottest: {hottest:.2f}°C")
print(f"Coldest: {coldest:.2f}°C")
```


## Analysis

The dataset contains 6 temperature readings ranging from -5.00°C to 38.50°C.

The average temperature was 17.08°C, which falls in the Warm category.

The lowest reading was below freezing and could create icy conditions.

Two readings were 30°C or higher, showing that some days were very hot.
